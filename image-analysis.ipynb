```python
from transformers import pipeline

nlp = pipeline(
    "document-question-answering",
    model="impira/layoutlm-document-qa",
)

image_url = os.path.join("images","image1.png")
image = Image.open(image_url)
```

```python
date = nlp(
    image,
    "What is the selected date in the screen-time widget?"
)
# {'score': 0.9943977, 'answer': 'us-001', 'start': 15, 'end': 15}

time = nlp(
    image,
    "What is the total amount of screen time for the selected day?"
)
# {'score': 0.9912159, 'answer': '$1,000,000,000', 'start': 97, 'end': 97}

data = nlp(
    image,
    "Estimate the screen-time for each of the 24 hours based on the bar chart."
)
# {'score': 0.59147286, 'answer': '$ 3,750', 'start': 19, 'end': 20}
```
